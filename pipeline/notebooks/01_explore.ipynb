Load a dataset, plot it, and check the trend.

```python
import json
from pathlib import Path
import matplotlib.pyplot as plt
import sys
sys.path.append('..')
from src.trend import analyze, deseasonalize
```

```python
dataset = json.loads(Path('../../web/public/datasets/phangnga-t2m.json').read_text(encoding='utf-8'))
series = dataset['series']
values = [p['value'] for p in series]
```

```python
fig, ax = plt.subplots(figsize=(10, 3))
ax.plot([p['t'] for p in series], values)
ax.set_xticks(ax.get_xticks()[::24])
ax.set_title(f"{dataset['variable']['label']} - {dataset['region']['name']}")
plt.show()
```

```python
result = analyze(deseasonalize(values))
print(f'Slope: {result.slope_per_step * 120:+.3f} per decade')
print(f'p-value: {result.p_value:.4f}, significant: {result.significant}')
```
